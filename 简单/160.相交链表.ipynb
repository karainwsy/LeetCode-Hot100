```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        # 超时：
        # curA = headA
        # while curA:
        #     curB = curB
        #     while curB:
        #         if curA is not curB:
        #             curB = curB.next
        #     return curA
        #     curA = curA.next
        # return None

        pA = headA
        pB = headB
        while pA != pB:
            if pA:
                pA = pA.next
            else:
                pA = headB
            
            if pB:
                pB = pB.next
            else:
                pB = headA
        return pA
```
