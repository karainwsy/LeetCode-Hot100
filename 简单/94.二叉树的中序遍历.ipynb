```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: TreeNode | None) -> list[int]:
        # 递归：
        # result = []
        # def dfs(node):
        #     if node is None:
        #         return
        #     dfs(node.left)
        #     result.append(node.val)
        #     dfs(node.right)
        # dfs(root)
        # return result
        result = []
        stack = []
        cur = root
        while cur or stack:
            while cur:
                stack.append(cur)
                cur = cur.left
            cur = stack.pop()
            result.append(cur.val)
            cur = cur.right
        return result
        
                
```
